```python
from datasets import load_dataset

ds = load_dataset("kanishka089/RainPrediction")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
# Assuming the dataset has a 'train' split and converting it to a pandas DataFrame
df = ds['train'].to_pandas()

# Display the first few rows and columns to verify
display(df.head())
display(df.columns)
```

Now, let's create a scatter plot using 'MinTemp' and 'MaxTemp' features. These are common features in weather-related datasets and often show a positive correlation.

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Humidity', y='Temperature', data=df)
plt.title('Scatter Plot of Humidity vs. Temperature')
plt.xlabel('Humidity (%)')
plt.ylabel('Temperature (°C)')
plt.grid(True)
plt.show()
```

Let's create a bar plot to see the relationship between 'Rain' and 'Temperature'.

```python
# Calculate the average temperature for each 'Rain' category
average_temp_by_rain = df.groupby('Rain')['Temperature'].mean().reset_index()

plt.figure(figsize=(8, 6))
sns.barplot(x='Rain', y='Temperature', data=average_temp_by_rain)
plt.title('Average Temperature by Rain Occurrence')
plt.xlabel('Rain (0 = No Rain, 1 = Rain)')
plt.ylabel('Average Temperature (°C)')
plt.xticks([0, 1], ['No Rain', 'Rain'])
plt.grid(axis='y')
plt.show()
```

Let's visualize the distribution of 'Temperature' using a histogram.

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['Temperature'], bins=20, kde=False)
plt.title('Distribution of Temperature')
plt.xlabel('Temperature (°C)')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['Temperature'], bins=20, kde=True)
plt.title('Distribution of Temperature')
plt.xlabel('Temperature (°C)')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)
plt.show()
```
