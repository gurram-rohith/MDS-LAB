### Importing a CSV File into a DataFrame
You can use `pd.read_csv()` to load data from a CSV file. Replace `'your_file.csv'` with the actual path to your CSV file.

```python
try:
    df_csv = pd.read_csv('/content/sample_data/california_housing_test.csv')
    print("DataFrame loaded successfully from 'california_housing_test.csv':")
    display(df_csv)
except FileNotFoundError:
    print("Error: The specified CSV file was not found. Please provide the correct path.")
except Exception as e:
    print(f"An error occurred: {e}")
```

```python
df_2=df_csv['median_income'].rank(ascending=False)
```

```python
df_2.head()
```

```python
print('Ranks sorted in descending order (highest median_income first):')
display(df_2.sort_values(ascending=True).head())
```

```python
print('Full ranks for median_income:')
display(df_2)
```

```python
df_csv['median_income_rank'] = df_2
print('df_csv with new median_income_rank column:')
display(df_csv.head())
```

```python
print("Original columns:")
display(df_csv.columns)


df_renamed_single = df_csv.rename(columns={'median_income': 'MedianIncome'})
print("\nDataFrame after renaming 'median_income' to 'MedianIncome':")
display(df_renamed_single.head())


df_renamed_multiple = df_csv.rename(columns={
    'housing_median_age': 'HousingAge',
    'total_rooms': 'TotalRooms',
    'total_bedrooms': 'TotalBedrooms'
})
print("\nDataFrame after renaming multiple columns:")
display(df_renamed_multiple.head())
```
