### Demonstrating Universal Functions (ufuncs) and Mathematical Operations

```python
import numpy as np

# Create some example NumPy arrays
arr1 = np.array([1, 2, 3, 4, 5])
arr2 = np.array([6, 7, 8, 9, 10])
scalar = 2

print("Array 1:", arr1)
print("Array 2:", arr2)
print("Scalar:", scalar)
```

#### Element-wise operations (ufuncs)

```python
# Addition
sum_arr = np.add(arr1, arr2) # Equivalent to arr1 + arr2
print("Addition (arr1 + arr2):"), display(sum_arr)

# Subtraction
diff_arr = np.subtract(arr1, arr2) # Equivalent to arr1 - arr2
print("Subtraction (arr1 - arr2):"), display(diff_arr)

# Multiplication
prod_arr = np.multiply(arr1, arr2) # Equivalent to arr1 * arr2
print("Multiplication (arr1 * arr2):"), display(prod_arr)

# Division
div_arr = np.divide(arr1, arr2) # Equivalent to arr1 / arr2
print("Division (arr1 / arr2):"), display(div_arr)

# Exponentiation
pow_arr = np.power(arr1, scalar) # Equivalent to arr1 ** scalar
print("Power (arr1 ** scalar):"), display(pow_arr)
```

```python
arr_math = np.array([0, np.pi/2, np.pi, 2, 4, 9])
print("Array for mathematical ufuncs:", arr_math)

# Square root
sqrt_arr = np.sqrt(arr_math)
print("Square root:"), display(sqrt_arr)

# Sine
sin_arr = np.sin(arr_math)
print("Sine:"), display(sin_arr)

# Cosine
cos_arr = np.cos(arr_math)
print("Cosine:"), display(cos_arr)

# Exponential
exp_arr = np.exp(arr_math)
print("Exponential:"), display(exp_arr)

# Natural logarithm (ln)
# Note: log of 0 or negative numbers will result in inf or NaN
log_arr = np.log(arr_math[arr_math > 0]) # Avoid log of non-positive numbers
print("Natural Logarithm (for positive elements):"), display(log_arr)
```

#### Aggregation methods

```python
agg_arr = np.array([1, 5, 2, 8, 3, 9, 4, 7, 6])
print("Array for aggregation:", agg_arr)

# Sum of all elements
sum_val = np.sum(agg_arr)
print(f"Sum of elements: {sum_val}")

# Minimum element
min_val = np.min(agg_arr)
print(f"Minimum element: {min_val}")

# Maximum element
max_val = np.max(agg_arr)
print(f"Maximum element: {max_val}")

# Mean (average)
mean_val = np.mean(agg_arr)
print(f"Mean of elements: {mean_val:.2f}")

# Standard Deviation
std_val = np.std(agg_arr)
print(f"Standard Deviation: {std_val:.2f}")
```

### Statistical and Comparison Operations on DataFrames

```python
import pandas as pd
import numpy as np

# --- Data Loading or Creation --- #
csv_file_path = '/tmp/my_sample_data.csv' # Placeholder CSV path

try:
    # Attempt to load the CSV file
    df = pd.read_csv(csv_file_path)
    print(f"Successfully loaded data from {csv_file_path}")
except FileNotFoundError:
    # If the file doesn't exist, create a sample DataFrame
    print(f"CSV file not found at {csv_file_path}. Creating a sample DataFrame.")
    data = {
        'A': np.random.randint(1, 100, 10),
        'B': np.random.uniform(10.0, 50.0, 10),
        'C': np.random.choice(['X', 'Y', 'Z'], 10),
        'D': np.random.rand(10) * 100
    }
    df = pd.DataFrame(data)

print("\nInitial DataFrame:")
display(df.head())
```

#### Statistical Operations

```python
# Mean of numerical columns
print("Mean of each numerical column:")
display(df.mean(numeric_only=True))

# Median of numerical columns
print("\nMedian of each numerical column:")
display(df.median(numeric_only=True))

# Standard Deviation of numerical columns
print("\nStandard Deviation of each numerical column:")
display(df.std(numeric_only=True))

# Minimum value of each numerical column
print("\nMinimum value of each numerical column:")
display(df.min(numeric_only=True))

# Maximum value of each numerical column
print("\nMaximum value of each numerical column:")
display(df.max(numeric_only=True))

# Sum of a specific column, e.g., 'A'
print("\nSum of column 'A':", df['A'].sum())

# Describe (summary statistics) of the DataFrame
print("\nSummary statistics for numerical columns:")
display(df.describe())
```

#### Comparison Operations

```python
# Compare a column to a scalar value
print("Rows where 'A' is greater than 50:")
display(df[df['A'] > 50])

# Compare a column to another column
print("\nRows where 'A' is greater than 'D':")
display(df[df['A'] > df['D']])

# Check for equality
print("\nRows where 'C' is equal to 'Y':")
display(df[df['C'] == 'Y'])

# Check for inequality
print("\nRows where 'C' is not equal to 'X':")
display(df[df['C'] != 'X'])

# Using `isin` for multiple values comparison
print("\nRows where 'C' is either 'X' or 'Z':")
display(df[df['C'].isin(['X', 'Z'])])

# Boolean indexing with multiple conditions
print("\nRows where 'A' > 50 AND 'B' < 30:")
display(df[(df['A'] > 50) & (df['B'] < 30)])

print("\nRows where 'A' > 80 OR 'D' < 20:")
display(df[(df['A'] > 80) | (df['D'] < 20)])
```

```python
import cv2
import numpy as np
import requests
import os

def download_image_to_array(url):
    """Download image from URL directly into memory (no file saved)."""
    # Add a User-Agent header to mimic a browser, which can sometimes bypass 403/429 errors
    headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/100.0.4896.127 Safari/537.36'} # More generic User-Agent
    response = requests.get(url, headers=headers, timeout=15) # Increased timeout
    response.raise_for_status()
    img_array = np.frombuffer(response.content, np.uint8)
    img = cv2.imdecode(img_array, cv2.IMREAD_COLOR)
    if img is None:
        raise ValueError("Could not decode image from URL")
    return img


def extract_iris(url, output_dir="iris_output"):
    """
    Download an animal image from a URL, detect eyes, then extract
    the iris/pupil region from each detected eye.
    """
    os.makedirs(output_dir, exist_ok=True)

    img = download_image_to_array(url)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    gray_eq = cv2.equalizeHist(gray)

    cascade_path = cv2.data.haarcascades + "haarcascade_eye.xml" # Corrected typo in previous step
    eye_cascade = cv2.CascadeClassifier(cascade_path)
    if eye_cascade.empty():
        raise IOError("Could not load haarcascade_eye.xml")

    eyes = eye_cascade.detectMultiScale(
        gray_eq, scaleFactor=1.03, minNeighbors=3, minSize=(15, 15) # Adjusted parameters for more lenient detection
    )

    if len(eyes) == 0:
        print("No eyes detected. Try adjusting scaleFactor/minNeighbors, "
              "or crop the eye region manually.")
        return []

    saved_paths = []
    vis_img = img.copy()

    for i, (x, y, w, h) in enumerate(eyes):
        # Crop the eye region (with slight padding)
        pad = int(0.1 * w)
        x1, y1 = max(0, x - pad), max(0, y - pad)
        x2, y2 = min(img.shape[1], x + w + pad), min(img.shape[0], y + h + pad)
        eye_roi = img[y1:y2, x1:x2]
        eye_gray = cv2.cvtColor(eye_roi, cv2.COLOR_BGR2GRAY)
        eye_gray = cv2.medianBlur(eye_gray, 5)  # reduce noise for circle detection

        # Detect circles (iris/pupil) within the eye region
        circles = cv2.HoughCircles(
            eye_gray,
            cv2.HOUGH_GRADIENT,
            dp=1.2,
            minDist=eye_gray.shape[0] // 8, # Adjusted minDist to be more lenient
            param1=25, # Lowered param1 further
            param2=12, # Lowered param2 further
            minRadius=eye_gray.shape[0] // 10, # Kept minRadius
            maxRadius=eye_gray.shape[0] // 3 # Kept maxRadius
        )

        if circles is None:
            print(f"Eye {i+1}: no iris circle found, saving full eye crop instead.")
            out_path = os.path.join(output_dir, f"eye_{i+1}_full.png")
            cv2.imwrite(out_path, eye_roi)
            saved_paths.append(out_path)
            continue

        circles = np.round(circles[0, :]).astype("int")
        # Take the most prominent circle (first one, usually strongest match)
        cx, cy, r = circles[0]

        # Crop a square region around the iris circle
        ix1, iy1 = max(0, cx - r), max(0, cy - r)
        ix2, iy2 = min(eye_roi.shape[1], cx + r), min(eye_roi.shape[0], cy + r)
        iris_crop = eye_roi[iy1:iy2, ix1:ix2]

        out_path = os.path.join(output_dir, f"iris_{i+1}.png")
        cv2.imwrite(out_path, iris_crop)
        saved_paths.append(out_path)

        # Draw detected circle on visualization (mapped back to full image coords)
        cv2.circle(vis_img, (x1 + cx, y1 + cy), r, (0, 255, 0), 2)
        cv2.circle(vis_img, (x1 + cx, y1 + cy), 2, (0, 0, 255), 3)

    vis_path = os.path.join(output_dir, "detected_iris_overlay.png")
    cv2.imwrite(vis_path, vis_img)

    print(f"Extracted {len(saved_paths)} iris/eye crop(s) into '{output_dir}/'")
    return saved_paths


if __name__ == "__main__":
    # Using a direct link to a raw image file from a public GitHub repository.
    url = "https://raw.githubusercontent.com/opencv/opencv/4.x/samples/data/baboon.jpg"
    extract_iris(url)
```

### Import CSV without Pandas and Statistical Computations

First, let's load the data from a CSV file into a basic Python list structure, without using the pandas library. We'll use the `csv` module from the standard library to achieve this.

```python
import csv
import os
import numpy as np # We can still use numpy for calculations after loading

csv_file_path = '/tmp/my_sample_data.csv'

# Ensure the sample CSV file exists for this demonstration
# If not, create a dummy one. This logic is similar to the pandas example.
if not os.path.exists(csv_file_path):
    print(f"CSV file not found at {csv_file_path}. Creating a sample CSV.")
    data_rows = [
        ['A', 'B', 'C', 'D'],
        [54, 39.24, 'Y', 62.35],
        [60, 15.23, 'X', 15.18],
        [53, 15.52, 'Y', 54.83],
        [76, 25.43, 'X', 85.70],
        [31, 31.87, 'X', 22.56],
        [19, 31.83, 'Z', 66.78],
        [73, 33.70, 'Y', 54.26],
        [36, 25.29, 'Z', 44.88],
        [39, 24.12, 'Y', 55.43],
        [29, 30.70, 'Y', 30.77]
    ]
    with open(csv_file_path, 'w', newline='') as f:
        writer = csv.writer(f)
        writer.writerows(data_rows)


raw_data = []
with open(csv_file_path, 'r') as file:
    reader = csv.reader(file)
    header = next(reader) # Read header row
    for row in reader:
        raw_data.append(row)

print("Header:", header)
print("First 5 rows of raw data:")
for i in range(min(5, len(raw_data))):
    print(raw_data[i])
```

Now that we have the data loaded as a list of lists, let's perform some basic statistical computations on the numerical columns. We'll convert the relevant columns to numerical types (e.g., `float`) and then use Python's built-in functions or NumPy for calculations.

```python
# @title
# Identify numerical columns by their index
# Assuming columns 'A', 'B', 'D' are numerical

# Get indices of numerical columns from the header
num_cols_indices = [header.index('A'), header.index('B'), header.index('D')]
num_cols_names = ['A', 'B', 'D']

# Extract numerical data and convert to float
numerical_data = []
for row in raw_data:
    num_row = []
    for idx in num_cols_indices:
        try:
            num_row.append(float(row[idx]))
        except ValueError:
            num_row.append(np.nan) # Handle non-numeric values
    numerical_data.append(num_row)

# Convert to a NumPy array for easier statistical operations
num_array = np.array(numerical_data)

print("Statistical Computations (using NumPy on extracted data):\n")

for i, col_name in enumerate(num_cols_names):
    col_data = num_array[:, i] # Select column
    col_data = col_data[~np.isnan(col_data)] # Remove NaNs for calculations

    if len(col_data) > 0:
        print(f"--- Column '{col_name}' ---")
        print(f"  Mean: {np.mean(col_data):.2f}")
        print(f"  Median: {np.median(col_data):.2f}")
        print(f"  Min: {np.min(col_data):.2f}")
        print(f"  Max: {np.max(col_data):.2f}")
        print(f"  Standard Deviation: {np.std(col_data):.2f}")
    else:
        print(f"--- Column '{col_name}' ---")
        print("  No valid numerical data to compute statistics.")
    print()
```

### Convert Image to NumPy Array

We can use the `cv2.imread()` function to load an image from a file path directly into a NumPy array. Each pixel's color information (e.g., BGR for color images) will be stored as elements in the array.

```python
import cv2
import numpy as np
from IPython.display import display, HTML, Image # Import Image class

# Path to one of the previously extracted iris images
image_path = 'iris_output/detected_iris_overlay.png'

# Load the image into a NumPy array
image_array = cv2.imread(image_path)

if image_array is not None:
    print(f"Successfully loaded image from: {image_path}")
    print(f"Image type: {type(image_array)}")
    print(f"Image shape (height, width, channels): {image_array.shape}")
    print(f"Image data type: {image_array.dtype}")

    # Display the first few rows and columns of the array to show its structure
    print("\nFirst 5x5x3 pixels (BGR values):")
    display(image_array[:5, :5, :])

    # Display the image as a visual check using IPython.display.Image
    print("\nOriginal Image:")
    display(Image(filename=image_path))
else:
    print(f"Error: Could not load image from {image_path}. Please check the path.")
```
